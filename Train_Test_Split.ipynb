Train Test Split Using Sklearn

```python
features = dataframe.drop('Outcome', axis=1) # This means we are dropping the column named Outcome. Here, it looks for columns if we specify axis=1, because, drop() defaults to axis=0 and searches in rows
target = dataframe['Outcome']
# Features are inputs and target is output
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,y_train,x_test,y_test = train_test_split(features,target,test_size=0.3)
```
